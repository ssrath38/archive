We use distribution plot to visualize the distribution of quantitative data.
Let's discuss some plots that allow us to visualise the distribution of a data set. These plots are:
- distplot/histplot
- jointplot
- pair plot
- rugplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = sns.load_dataset('tips')
df
```

```python
df['size'].unique()
```

```python
# distplot
sns.displot(df['total_bill'])
```

```python
# sns.displot(df['total_bill'], bins=100) # default bins=20
plt.subplot(1,2,1)
sns.histplot(df['total_bill'], bins=20, kde=True) # same as displot
plt.subplot(1,2,2)
sns.histplot(df['tip'], kde=True)
```

```python
# jointplot
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind='scatter')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind='kde')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind='hex')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind='reg')
```

```python
# pair plot
sns.pairplot(df,hue='sex', palette='rainbow')
```

```python
sns.pairplot(df,hue='size', palette='rainbow')
```

```python
# rug plot
sns.rugplot(df['total_bill'])
```
