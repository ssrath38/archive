### Dataframe

```python
import numpy as np
import pandas as pd
```

**Creating a Dataframe**

```python
data = {
    "Name": ["John", "Anna", "Peter", "Linda"],
    "Age": [28, 34, 29, 42],
    "City": ["New York", "Paris", "Berlin", "London"],
    "Salary": [65000, 70000, 62000, 85000],
}

df = pd.DataFrame(data)
```

```python
df
```

```python
data_list = [
    ["John", 28, "New York", 65000],
    ["Anna", 34, "Paris", 70000],
    ["Peter", 29, "Berlin", 62000],
    ["Linda", 42, "London", 85000],
]
df2 = pd.DataFrame(data_list)
```

```python
df2
```

```python
columns = ['Name','Age','City','Salary']
df2 = pd.DataFrame(data_list,columns=columns)
df2
```

**Selection and Indexing of Columns**

```python
df2[["Name","City"]]
```

**Creating a new column**

```python
df2["Designation"] = ["Doctor",'Eng.','Doctor','Eng.']
```

```python
df2
```

**Removing Columns and rows**

```python
# Removing rows
df2.drop(0,axis=0) # default is axis = 0
```

```python
# Removing columns
df2.drop('Name',axis=1) # the main dataframe is not changed it is a copy which is changed
df2.drop('Name',axis=1,inplace=True) # permanently delete
df2.drop(['Name','City'],axis=1) # select 2 column and delete
```

**Selecting rows**

```python
df2.loc[[0,1]]
```

```python
df2.iloc[3] # index location
```

**Selecting Subsets of Rows and Columns**

```python
df2.loc[[0,1]][["City",'Salary']]
```

```python
df2
```

```python
df2.loc[[2,3]][['Name','Age']]
```

**Conditional Selection**

```python
# Select only those people whose age is above 30
df2[df2['Age']>30]
```

```python
# Select only those people whose age is above 30 and from Paris
df2[(df2['Age']>30)&(df2['City']=='Paris')]
```
