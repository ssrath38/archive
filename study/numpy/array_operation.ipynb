**Matrix operations**

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
```

```python
A*B
```

```python
A@B
```

```python
np.dot(A,B) # same as A@B
```

#### Advance Array Manipulation

**Stacking Array**

```python
# stacking cant be done in 2 matrix it is possible in only 2
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
# vertical stack
np.vstack((a,b))
```

```python
# horizontal stack
np.hstack((a,b))
```

```python
# column stack
np.column_stack((a,b))
```

**Splitting Array**

```python
c = np.arange(16).reshape(4,4)
c
```

```python
# horizontal split
np.hsplit(c,2)
# np.hsplit(c,3) # Error
```

```python
# Vertical split
np.vsplit(c,2)
np.vsplit(c,4)
# np.vsplit(c,3) # error
```
