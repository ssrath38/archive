```python
import numpy as np
```

**NUMPY INDEXING AND SLICING OF VECTORS**

```python
arr = np.arange(11,21)
arr
```

```python
arr[6]
```

```python
arr[1:5] # last index is excluded
arr[:5] # default [0:5]
arr[:] # whole array
arr[3:] # end is last
arr[3::2] # start:end:step
arr_slice = arr[::-1] # reverse the array
arr_slice
```

**NUMPY INDEXING AND SLICING OF MATRIX**

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[1] # get the row
```

```python
arr[:,2] # get the column
```

```python
arr[0, 4] # row, column
```

```python
arr[3:,3:]
```

**BOOLEAN INDEXING**

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr % 2 == 0
bool_index
```

```python
arr = arr[bool_index] # the array size and bool_index size should be the same
arr
```
