**Array generation function**

```python
arr = np.arange(1,11,2) # arange(start, stop, step) stop not included
arr
```

```python
arr = np.zeros (6)
arr
```

```python
arr = np.zeros((4,8)) #(rows, columns)
arr
```

```python
arr = np.ones(6)
arr
```

```python
arr = np.ones((3,3))
arr
```

```python
arr = np.linspace(1,5,3) #linear spacing
arr
```

```python
arr.min()
```

```python
arr.max()
```

```python
arr.sum()
```

```python
arr = arr.reshape((6,5)) # note the size of the array and the reshaped array should be same
# resizing means save the reshaped array in a variable so that the array will saved in that size
arr
```
